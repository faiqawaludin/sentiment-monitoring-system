```python
import sys
import os
import pandas as pd
from sqlalchemy import Column, Integer, String, Text, TIMESTAMP
from sqlalchemy.orm import declarative_base
from datetime import datetime
```

```python
sys.path.append(os.path.abspath('../src'))
from utils.db import get_db_engine
```

```python
engine = get_db_engine()
print("✅ Koneksi DB Berhasil.")
```

```python
Base = declarative_base()

class NewsRaw(Base):
    __tablename__ = 'news_raw'

    id = Column(Integer, primary_key=True, autoincrement=True)
    scraped_at = Column(TIMESTAMP, default=datetime.now)
    keyword = Column(String, nullable=False) # Kata kunci pencarian
    source = Column(String)
    title = Column(Text, nullable=False)
    url = Column(Text, nullable=False, unique=True) # Unique agar tidak ada berita kembar
    published_date = Column(String)
    content_snippet = Column(Text)
```

```python
try:
    Base.metadata.create_all(engine)
    print(" Tabel 'news_raw' berhasil dibuat di PostgreSQL!")
except Exception as e:
    print(f" Gagal: {e}")
```
