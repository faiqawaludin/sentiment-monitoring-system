```python
from sqlalchemy import create_engine, Column, Integer, String, Boolean, TIMESTAMP
from sqlalchemy.orm import sessionmaker, declarative_base
from datetime import datetime
import pandas as pd
```

```python
# --- KONFIGURASI KONEKSI ---
# Format: postgresql://user:password@host:port/database
DB_URL = "postgresql://remosy_user:remosy_password@localhost:5434/remosy_dw"
```

```python
try:
    engine = create_engine(DB_URL)
    connection = engine.connect()
    print(" Berhasil koneksi ke PostgreSQL!")
    connection.close()
except Exception as e:
    print(f" Gagal koneksi: {e}")
```

```python
Base = declarative_base()

class ConfigKeywords(Base):
    __tablename__ = 'config_keywords'

    id = Column(Integer, primary_key=True, autoincrement=True)
    keyword = Column(String, nullable=False)
    category = Column(String)
    is_active = Column(Boolean, default=True)
    created_at = Column(TIMESTAMP, default=datetime.now)
```

```python
try:
    Base.metadata.create_all(engine)
    print(" Tabel 'config_keywords' berhasil dibuat (atau sudah ada).")
except Exception as e:
    print(f" Error create table: {e}")
```

```python
Session = sessionmaker(bind=engine)
session = Session()

if session.query(ConfigKeywords).count() == 0:
    data_awal = [
        ConfigKeywords(keyword="Universitas Singaperbangsa Karawang", category="Institusi"),
        ConfigKeywords(keyword="UNSIKA", category="Institusi"),
        ConfigKeywords(keyword="Rektor UNSIKA", category="Tokoh")
    ]
    session.add_all(data_awal)
    session.commit()
    print(" Data default berhasil dimasukkan.")
else:
    print("ℹ Data sudah ada, skip insert.")

session.close()
```

```python
print("\n--- Isi Tabel di PostgreSQL ---")
df = pd.read_sql("SELECT * FROM config_keywords", engine)
display(df)
```
